**Άσκηση 4**

Έστω ότι σχεδιάζετε ένα Bayesian σύστημα ταξινόμησης σε 2 κλάσεις ω = {ω1, ω2} , και
καταλήγετε πως η πιθανότητα σφάλματος του συστήματος εξαρτάται από την a priori
πιθανότητα P1 της κλάσης ω1 σύμφωνα με τη σχέση:

Perror(P1) = (P1^2)*(1-P1) 0<=P1<=1

```python
# Κελί 1
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,1,100)    # Το x θα είναι οι τιμές του άξονα x, όπου σύμφωνα με την εκφώνηση θα αντιστοιχεί στο P1
print(x)                    # Λόγω πράξεων στο δυαδικό σύστημα απο μεριάς του υπολογιστή χάνεται η ακρίβεια (Το 0.01 δεν μπορει να αναπαρασταθει στο δυαδικο)
x = np.round(x, 2)          # Η στρογγυλοποίηση λύνει το "πρόβλημα"
print(x)
Pe = [(i**2)*(1-i) for i in x] #Το Perror

plt.plot(x, Pe)           # Η γραφική παράσταση του σφάλματος συναρτήσει του P1
plt.show()
```

```python
# To σφάλμα αν σχεδιάσουμε το σύστημα για το μέγιστο σφάλμα
Pe_max = max(Pe)      # Η μέγιστη τιμή του Perror. Εδώ δεν χρειάζεται να βρούμε που μηδενίζει η παράγωγος.
x_max = x[np.argmax(Pe)]  # Βρίσκω το index του πίνακα x (ή P1 από την εκφώνηση της άσκσησης) για τον οποίο υπολογίζεται το Pe_max.
print("Μέγιστο σφάλμα έχουμε για P1 = ", x_max)
print("και είναι ίσο με Pe(", x_max, ") = ", Pe_max)
```

```python
C_min = [i*0+Pe_max for i in x] # Ευθεία που αντιπροσωπέυει το σύστημα που σχεδιάστηκε ωστε να ελαχιστοποιηθεί το μέγιστο σφάλμα.
                                # Υπολογίστηκε με την μέθοδο Minimax στο report.
plt.plot(x, Pe)                 # Η γραφική παράσταση του σφάλματος συναρτήσει του P1
plt.plot(x, C_min)              # Η γραφική παράσταση της ευθείας που αντιπροσωπέυει το σύστημα συναρτήσει του P1
plt.show()                      # Τύπωση των 2 γραφικών παραστάσεων σε κοινό διάγραμμα
```

**Ερώτημα β:** Εάν ρυθμίσατε το σύστημά σας υποθέτοντας πιθανότητα P1=0.3, ποια θα είναι η
πιθανότητα σφάλματος εάν η πραγματική a priori πιθανότητα της κλάσης ω1 είναι P1=0.7

```python
A = 0.33 # Υπολογίστηκε με το χέρι στο report
B = -0.036 # Υπολογίστηκε με το χέρι στο report
```

```python
C = [i*A+B for i in x] # Η ευθεία του συστήματος αν έχει ρυθμιστεί για P1 = 0.3
plt.plot(x, Pe)       # Η γραφική παράσταση του σφάλματος συναρτήσει του P1
plt.plot(x, C)        # Η γραφική παράσταση της ευθείας που αντιπροσωπέυει το σύστημα
                      # που σχεδιάστηκε, θεωρώντας ότι το P1=0.3, συναρτήσει του P1

plt.show()            # Τύπωση των 2 γραφικών παραστάσεων σε κοινό διάγραμμα
```

```python
x_0_7 = np.where(x==0.7)[0][0] # Υπολογίζω την τιμή της πιθανότητας σφάλματος αν τελικά το P1 = 0.7
print(C[x_0_7])
```
